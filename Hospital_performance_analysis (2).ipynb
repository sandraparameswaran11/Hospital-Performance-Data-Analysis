#**Handling Column Name Issues**

*   Checking Column Names, Striping Extra Spaces, Standardizing Case, Replacing Special Characters, Renaming Specific Columns




```python
# Clean and standardize column names
df.columns = df.columns.str.strip().str.lower().str.replace(r'\s+', '_', regex=True)
df.columns = df.columns.str.replace('\xa0', '', regex=True)

# Rename specific columns for clarity
df.rename(columns={
    'charges_amt': 'treatment_cost',
    'insurance_amt': 'insurance_covered',
    'patient_satisfaction': 'satisfaction_score',
    'admissiondate': 'admission_date',
    'dischargedate': 'discharge_date',
    'lengthofstay': 'length_of_stay',
    'patientid': 'patient_id',
    'patientname': 'patient_name',
    'doctorfee': 'doctor_fee'
}, inplace=True)

print(df.columns)



```

#**Change Data Types**
*  Several columns were initially stored as object (text) data types and required conversion before analysis.
*  The admission_date and discharge_date columns were converted to datetime format using pd.to_datetime(errors='coerce').
*  Validation confirmed that 0 invalid admission dates and 0 invalid discharge dates were present after conversion.
*  Financial columns (treatment_cost, insurance_covered, patient_bill, medicine_cost, test_cost, and doctor_fee) were cleaned by removing currency symbols and commas before being converted to numeric format using pd.to_numeric(errors='coerce').
*   Additional numerical columns (age, wait_time_minutes, and satisfaction_score) were also converted to numeric format.
*   Any invalid values encountered during conversion were converted to missing values (NaN) and subsequently filled using the median value of the respective column.
*   After conversion, all financial and operational variables were successfully stored in appropriate numeric formats, while admission and discharge dates were stored as datetime variables.








```python
# Convert date columns to datetime
df['admission_date'] = pd.to_datetime(df['admission_date'], errors='coerce')
df['discharge_date'] = pd.to_datetime(df['discharge_date'], errors='coerce')

# Check for invalid (NaT) dates after conversion
print("Invalid admission dates:", df['admission_date'].isna().sum())
print("Invalid discharge dates:", df['discharge_date'].isna().sum())

# Normalize department names
df['department'] = df['department'].astype(str).str.strip().str.title()

# Convert all financial columns to numeric
financial_cols = [
    'treatment_cost', 'insurance_covered', 'patient_bill',
    'medicine_cost', 'test_cost', 'doctor_fee'
]

for col in financial_cols:
    # Remove currency symbols and commas before converting
    df[col] = df[col].replace('[₹,]', '', regex=True)
    df[col] = pd.to_numeric(df[col], errors='coerce')

# Convert other numeric columns
numeric_cols = ['age', 'wait_time_minutes', 'satisfaction_score']
for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')

# Check for invalid numeric values after conversion
print("Invalid numeric values after conversion:")
print(df[financial_cols + numeric_cols].isna().sum())

# Handle missing values created by conversion
for col in financial_cols + numeric_cols:
    df[col] = df[col].fillna(df[col].median())

print(df.dtypes)

```

#**Handling Missing Values**
   
*   Missing values were identified using isnull().sum().
*  Categorical columns (city, department, gender, treatment_type, payment_method, doctor_name, hospital_name, admission_type, diagnosis, and bed_type) were cleaned and missing values were filled using the mode of each column.
*   Numerical columns (age, treatment_cost, insurance_covered, patient_bill, medicine_cost, test_cost, doctor_fee, wait_time_minutes, and satisfaction_score) were converted to numeric format and missing values were filled using the median of the respective column.
*   Text values such as "nan" in categorical columns were first converted to actual missing values (NaN) before imputation.
*   After applying the missing-value treatment, the dataset was verified and all missing values were successfully eliminated (Total Missing Values = 0).
*  Data type validation confirmed that the dataset contained appropriate object, numeric, and datetime variables after preprocessing.




```python
df.isnull().sum()
```

```python

import pandas as pd

# Handle categorical columns
categorical_cols = [
    'city', 'department', 'gender', 'treatment_type',
    'payment_method', 'doctor_name', 'hospital_name',
    'admission_type', 'diagnosis', 'bed_type'
]

for col in categorical_cols:
    df[col] = df[col].astype(str).str.strip().str.lower()
    df[col] = df[col].replace('nan', pd.NA).fillna(df[col].mode()[0])

# Handle numeric columns
numeric_cols = [
    'age', 'treatment_cost', 'insurance_covered', 'patient_bill',
    'medicine_cost', 'test_cost', 'doctor_fee',
    'wait_time_minutes', 'satisfaction_score'
]

for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')  # convert to numeric
    df[col] = df[col].fillna(df[col].median())         # fill missing with median

# Verify cleaning results
print("Missing values after cleaning:\n", df.isnull().sum())
print("Total missing values:", df.isnull().sum().sum())
print("\nData types after cleaning:\n", df.dtypes)


```

#**Handling Inconsistent Data**

*  Duplicate records were checked using the duplicated() function and removed where necessary to ensure data consistency.
*  Categorical variables (gender, department, payment_method, doctor_name, city, hospital_name, admission_type, diagnosis, treatment_type, readmission_flag, and bed_type) were standardized by converting values to lowercase and removing unnecessary spaces.
*  Variations and spelling inconsistencies in department names were corrected using a mapping dictionary.
*  Numerical columns (age and satisfaction_score) were validated for acceptable value ranges. Age values were restricted to 0-120, and satisfaction scores were limited to a scale of 1-10 using clipping.
* Date columns (admission_date and discharge_date) were validated to ensure logical consistency. Records with discharge dates earlier than admission dates were identified and removed.
* Dataset integrity was verified by checking duplicate patient IDs, data types, non-null counts, and unique values for each column.







```python
#Handling Inconsistent Data

# Check for duplicate records
duplicates = df[df.duplicated()]
print("Duplicate rows:", duplicates.shape[0])
df = df.drop_duplicates()

# Standardize categorical values
categorical_cols = [
    'gender', 'department', 'payment_method', 'doctor_name',
    'city', 'hospital_name', 'admission_type', 'diagnosis',
    'treatment_type', 'readmission_flag', 'bed_type'
]

for col in categorical_cols:
    df[col] = df[col].astype(str).str.lower().str.strip()

# Fix spelling variations and typos in department names
dept_map = {
    'cardio': 'cardiology',
    'cardiology ': 'cardiology',
    'cardiologgy': 'cardiology',
    'orthopedics': 'orthopaedics',
    'ortho': 'orthopaedics',
    'neuro': 'neurology',
    'neurologgy': 'neurology',
    'ent ': 'ent',
    'paediatrics': 'pediatrics',
    'pediatrics': 'pediatrics',
    'gynecology': 'gynaecology',
    'gynaecology': 'gynaecology'
}

df['department'] = df['department'].replace(dept_map)


# Check numeric outliers
print(df['age'].describe())
print(df['satisfaction_score'].describe())

# Validate age range
df = df[(df['age'] >= 0) & (df['age'] <= 120)]

# Validate satisfaction scores
invalid_satisfaction = df[(df['satisfaction_score'] < 1) | (df['satisfaction_score'] > 10)]
print("Invalid satisfaction scores:", len(invalid_satisfaction))
df['satisfaction_score'] = df['satisfaction_score'].clip(1, 10)

# Check inconsistent dates
inconsistent_dates = df[
    (df['discharge_date'] < df['admission_date']) &
    df['discharge_date'].notna() & df['admission_date'].notna()
]
print("Rows with inconsistent dates:", inconsistent_dates.shape[0])

# Drop or correct inconsistent rows
df = df[df['discharge_date'] >= df['admission_date']]

# Verify dataset integrity
print("Duplicate patient IDs:", df['patient_id'].duplicated().sum())
print(df.info())
print(df.nunique())


```

* Outliers were treated using the capping (Winsorization) technique to preserve all records in the dataset.
*  Columns analyzed were age, treatment_cost, insurance_covered, wait_time_minutes, and satisfaction_score.
*  A total of 1 outlier was detected in age, 28 outliers in insurance_covered, and no outliers were detected in treatment_cost, wait_time_minutes, or satisfaction_score.













```python

import numpy as np

# Identify numeric columns
numeric_cols = ['age', 'treatment_cost', 'insurance_covered',
                'wait_time_minutes', 'satisfaction_score']

# Apply IQR method to each numeric column
for col in numeric_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[(df[col] < lower_bound) | (df[col] > upper_bound)]
    print(f"{col}: {outliers.shape[0]} outliers detected")

    # Option 1: Remove outliers
    # df = df[(df[col] >= lower_bound) & (df[col] <= upper_bound)]

    # Option 2: Cap outliers (recommended to preserve data)
    df[col] = np.where(df[col] < lower_bound, lower_bound,
                       np.where(df[col] > upper_bound, upper_bound, df[col]))

# Review numeric columns after cleaning
print(df[numeric_cols].describe())

```
